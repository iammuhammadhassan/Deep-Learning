```python
import numpy as np
import matplotlib.pyplot as plt

# XOR inputs
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

# XOR outputs
y = np.array([0, 1, 1, 0])

# Plot XOR points
plt.figure(figsize=(7, 5))

for i in range(len(X)):
    
    if y[i] == 0:
        plt.scatter(
            X[i][0],
            X[i][1],
            marker='o',
            s=200,
            label='Output 0' if i == 0 else ""
        )
        
    else:
        plt.scatter(
            X[i][0],
            X[i][1],
            marker='x',
            s=200,
            label='Output 1' if i == 1 else ""
        )

# Labels for each point
for i in range(len(X)):
    plt.annotate(
        f"({X[i][0]}, {X[i][1]}) → {y[i]}",
        (X[i][0], X[i][1]),
        xytext=(8, 8),
        textcoords="offset points"
    )

plt.xlabel("Input X1")
plt.ylabel("Input X2")

plt.title("XOR Gate - Non-Linearly Separable Problem")

plt.xlim(-0.2, 1.2)
plt.ylim(-0.2, 1.2)

plt.xticks([0, 1])
plt.yticks([0, 1])

plt.grid(True)
plt.legend()

plt.show()
```
