```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-10, 10, 1000)
```

Sigmoid function

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

```python
print(sigmoid(-10))
print(sigmoid(0))
print(sigmoid(10))
```

```python
plt.plot(x, sigmoid(x))

plt.xlabel("Input (x)")
plt.ylabel("Sigmoid Output")
plt.title("Sigmoid Activation Function")

plt.grid(True)
plt.show()
```

Tanh

```python
def tanh(x):
    return np.tanh(x)
```

```python
print(tanh(-10))
print(tanh(0))
print(tanh(10))
```

```python
plt.plot(x, tanh(x))

plt.xlabel("Input (x)")
plt.ylabel("Tanh Output")
plt.title("Tanh Activation Function")

plt.grid(True)
plt.show()
```

Relu

```python
def relu(x):
    return np.maximum(0, x)
```

```python
print(relu(-10))
print(relu(0))
print(relu(10))
```

```python
plt.plot(x, relu(x))

plt.xlabel("Input (x)")
plt.ylabel("ReLU Output")
plt.title("ReLU Activation Function")

plt.grid(True)
plt.show()
```

```python
def leaky_relu(x, alpha=0.01):
    return np.where(x > 0, x, alpha * x)
```

```python
print(leaky_relu(-10))
print(leaky_relu(0))
print(leaky_relu(10))
```

```python
plt.plot(x, leaky_relu(x))

plt.xlabel("Input (x)")
plt.ylabel("Leaky ReLU Output")
plt.title("Leaky ReLU Activation Function")

plt.grid(True)
plt.show()
```
