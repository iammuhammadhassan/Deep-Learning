```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-10, 10, 1000)
```

Sigmoid function

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

```python
print(sigmoid(-10))
print(sigmoid(0))
print(sigmoid(10))
```

```python
plt.plot(x, sigmoid(x))

plt.xlabel("Input (x)")
plt.ylabel("Sigmoid Output")
plt.title("Sigmoid Activation Function")

plt.grid(True)
plt.show()
```
