```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])
```

```python
y = np.array([0, 0, 0, 1])
```

```python
weights = np.zeros(2)
bias = 0

learning_rate = 0.1
epochs = 10
```

```python
def step_function(x):
    if x >= 0:
        return 1
    else:
        return 0
```

```python
for epoch in range(epochs):

    for i in range(len(X)):

        # Weighted sum
        z = np.dot(X[i], weights) + bias

        # Prediction
        prediction = step_function(z)

        # Error
        error = y[i] - prediction

        # Update weights
        weights = weights + learning_rate * error * X[i]

        # Update bias
        bias = bias + learning_rate * error

    print("Epoch:", epoch + 1,
          "Weights:", weights,
          "Bias:", bias)
```

```python
print("\nAND Gate Results:")

predictions = []

for x in X:

    z = np.dot(x, weights) + bias
    prediction = step_function(z)

    predictions.append(prediction)

    print(x, "=>", prediction)
```

```python
accuracy = np.mean(predictions == y) * 100

print("\nAccuracy:", accuracy, "%")
```

```python

plt.figure(figsize=(7, 5))

for i in range(len(X)):

    if y[i] == 0:
        plt.scatter(X[i][0], X[i][1],
                    marker='o', s=150,
                    label='Output 0' if i == 0 else "")

    else:
        plt.scatter(X[i][0], X[i][1],
                    marker='x', s=150,
                    label='Output 1' if i == 3 else "")
```

```python
x_values = np.linspace(-0.2, 1.2, 100)

if weights[1] != 0:
    y_values = -(weights[0] * x_values + bias) / weights[1]
    plt.plot(x_values, y_values,
             label='Decision Boundary')

plt.xlabel("Input X1")
plt.ylabel("Input X2")
plt.title("AND Gate - Perceptron Visualization")
plt.xlim(-0.2, 1.2)
plt.ylim(-0.2, 1.2)
plt.grid(True)
plt.legend()

plt.show()
```
